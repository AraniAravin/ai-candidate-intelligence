```python
from sentence_transformers import SentenceTransformer
model = SentenceTransformer('all-MiniLM-L6-v2')
```

```python
job_description = "Python backend developer"

cv_snippets = [
    "Developed APIs using Python and FastAPI",
    "Experienced React frontend engineer",
    "Built machine learning pipelines with Python and scikit-learn",
    "5 years of Java backend development",
    "Skilled in Python, Django, and PostgreSQL for backend services",
    "Python backend"
]
```

```python
job_embedding = model.encode(job_description)
cv_embeddings = model.encode(cv_snippets)

print("Embedding shape:", job_embedding.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarities = cosine_similarity([job_embedding], cv_embeddings)[0]

for snippet, score in zip(cv_snippets, similarities):
    print(f"{score:.4f}  —  {snippet}")
```
